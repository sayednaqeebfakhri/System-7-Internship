#### Feature engineering
Feature engineering means creating, transforming, or selecting features (input columns) to help a machine learning model learn better.

```python
import pandas as pd

data = {
    "study_hours": [2, 5, 7, 3, 6],
    "attendance": [60, 80, 95, 65, 90],
    "marks": [45, 70, 90, 50, 85]
}

df = pd.DataFrame(data)

# Create a new feature
df["study_attendance_score"] = (
    df["study_hours"] * df["attendance"]
)

print(df)
```
