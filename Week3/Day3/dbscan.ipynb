#### DBSCAN
DBSCAN stands for Density-Based Spatial Clustering of Applications with Noise.

It is an unsupervised machine learning algorithm that groups data points based on their density.

Unlike K-Means, DBSCAN:
- Does not require to specify the number of clusters (K).
- Can identify clusters with irregular shapes.
- Can detect outliers and noise points.

#### Important concepts
DBSCAN mainly uses two parameters:

1. eps (Epsilon)

The maximum distance between two points for them to be considered neighbors.

For example, eps=0.5 means points within a distance of 0.5 can be considered neighbors.

2. min_samples

The minimum number of points required within a point's eps neighborhood for that point to be considered a core point. The point itself is included in this count.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Load dataset
iris = load_iris()
X = iris.data

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create DBSCAN model
model = DBSCAN(
    eps=0.5,
    min_samples=5
)

# Fit and predict clusters
clusters = model.fit_predict(X_scaled)

print("Cluster assignments:")
print(clusters)

# Visualize clusters using first two features
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap="viridis"
)

plt.xlabel("Sepal Length (scaled)")
plt.ylabel("Sepal Width (scaled)")
plt.title("DBSCAN Clustering on Iris Dataset")

plt.show()
```
