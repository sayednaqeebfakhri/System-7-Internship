#### Hierarchical Clustering
Hierarchical Clustering is an unsupervised machine learning algorithm that groups similar data points into clusters by building a hierarchy of clusters.

Two main approaches:
1. Agglomerative (bottom-up): Starts with every data point as its own cluster and gradually merges similar clusters.
2. Divisive (top-down): Starts with all data points in one cluster and gradually splits them into smaller clusters.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

# Load dataset
iris = load_iris()
X = iris.data

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create model
model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

# Fit and predict clusters
clusters = model.fit_predict(X_scaled)

print("Cluster assignments:")
print(clusters)

# Visualize clusters using first two features
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap="viridis"
)

plt.xlabel("Sepal Length (scaled)")
plt.ylabel("Sepal Width (scaled)")
plt.title("Hierarchical Clustering")

plt.show()
```
