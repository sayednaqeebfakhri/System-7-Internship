#### Feature Scaling
Feature scaling means transforming numerical features into comparable ranges.

##### StandardScaler

StandardScaler transforms data so that each feature has approximately:
- Mean = 0
- Standard deviation = 1

##### MinMaxScaler

MinMaxScaler usually transforms numerical values into a range between 0 and 1.

```python
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 35],
    "Salary": [30000, 50000, 80000, 60000, 90000],
    "City": ["Islamabad", "Lahore", "Karachi", "Islamabad", "Lahore"],
    "Education": ["BS", "MS", "BS", "PhD", "MS"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# standard scalar

from sklearn.preprocessing import StandardScaler

# Select numerical columns
numeric_data = df[["Age", "Salary"]]

# Create scaler
scaler = StandardScaler()

# Apply scaling
scaled_data = scaler.fit_transform(numeric_data)

# Convert into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=["Age", "Salary"]
)

print(scaled_df)
```

```python
# OneHotEncoder
from sklearn.preprocessing import OneHotEncoder

# Select categorical columns
categorical_data = df[["City", "Education"]]

# Create encoder
encoder = OneHotEncoder(sparse_output=False)

# Apply encoding
encoded_data = encoder.fit_transform(categorical_data)

# Get column names
encoded_columns = encoder.get_feature_names_out(["City", "Education"])

# Convert into DataFrame
encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoded_columns
)

print(encoded_df)
```

```python
# Combine numerical and categorical data

final_df = pd.concat([scaled_df, encoded_df],axis=1)

print(final_df)
```

#### Column Transformer

```python
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Create dataset
data = {
    "Age": [22, 25, 28, 30, 35],
    "Salary": [30000, 50000, 80000, 60000, 90000],
    "City": ["Islamabad", "Lahore", "Karachi", "Islamabad", "Lahore"],
    "Education": ["BS", "MS", "BS", "PhD", "MS"]
}

df = pd.DataFrame(data)

# Define numerical columns
numeric_features = ["Age", "Salary"]

# Define categorical columns
categorical_features = ["City", "Education"]

# Create ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),

        ("cat", OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ]
)

# Transform dataset
X_transformed = preprocessor.fit_transform(df)

print("Original Dataset:")
print(df)

print("\nTransformed Dataset:")
print(X_transformed)

print("\nTransformed Shape:")
print(X_transformed.shape)
```
