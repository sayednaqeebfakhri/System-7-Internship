#### Lists, tuples, dictionaries, sets

#### List

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits)
print(type(fruits))
```

```python
# Accessing list elements
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])   # apple
print(fruits[2])   # mango
print(fruits[-1])  # orange
```

```python
# List slicing
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
```

```python
# Adding and removing elements

fruits = ["apple", "banana"]

fruits.append("mango")
print(fruits)

fruits.insert(1, "orange")
print(fruits)

fruits.remove("banana")
print(fruits)

last_item = fruits.pop()
print(last_item)
print(fruits)
```

```python
# updating elements
fruits = ["apple", "banana", "mango"]
fruits[1] = "orange"
print(fruits)
```

```python
# looping through a list
students = ["Ali", "Sara", "Ahmed"]

for student in students:
    print(student)
```

```python
# List comprehension
squares = [x ** 2 for x in range(1, 6)]

print(squares)
```

#### Tuples

```python
# Creating a tuple
coordinates = (10, 20, 30)

print(coordinates)
print(coordinates[0])
print(type(coordinates))
```

```python
# Accessing tuple elements
colors = ("red", "green", "blue")

print(colors[1])
print(colors[-1])
```

```python
student = ("Naqeeb", 23, "AI")

name, age, field = student

print(name)
print(age)
print(field)
```

#### Dictionaries

```python
student = {
    "name": "Naqeeb",
    "age": 23,
    "field": "Artificial Intelligence",
    "cgpa": 3.6
}

print(student)
print(student["name"])
print(student["cgpa"])
```

```python
# Adding and updating values
student = {"name": "Naqeeb", "age": 23}

student["email"] = "naqeeb@example.com"
student["age"] = 24

print(student)
```

```python
# Removing dictionary entries
student = {
    "name": "Naqeeb",
    "age": 23,
    "field": "AI"
}

student.pop("age")

print(student)
```

```python
# loops through a dictionary
student = {
    "name": "Naqeeb",
    "age": 23,
    "field": "AI"
}

for key, value in student.items():
    print(key, ":", value)
```

#### Sets

```python
numbers = {10, 20, 30, 20, 10}

print(numbers)
```

```python
numbers = {10, 20, 30}

numbers.add(40)
numbers.update([50, 60])

numbers.remove(20)

print(numbers)
```

```python
# Set operations
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Union
union = set1 | set2
print("Union:", union)

# Intersection
intersection = set1 & set2
print("Intersection:", intersection)

# Difference
difference = set1 - set2
print("Difference:", difference)


```
