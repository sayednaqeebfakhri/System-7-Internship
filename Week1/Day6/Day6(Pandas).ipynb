#### Pandas

```python
import pandas as pd
```

##### Series — one-dimensional data

```python
marks = pd.Series([85, 92, 76, 88, 95])

print(marks)
print(marks.mean())
print(marks.max())
print(marks.min())
```

```python
marks = pd.Series(
    [85, 92, 76],
    index=["Ali", "Sara", "Ahmed"]
)

print(marks)
print(marks["Sara"])
```

##### DataFrames — tabular data

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [85, 92, 67, 95, 74],
    "City": ["Islamabad", "Lahore", "Karachi", "Islamabad", "Peshawar"]
}

df = pd.DataFrame(data)

print(df)
```

##### Filtering data

```python
passed = df[df["Marks"] >= 70]
passed
```

```python
high_marks = df[df["Marks"] >= 80]
print(high_marks)

older_students = df[df["Age"] > 21]
print(older_students)

top_students = df[
    (df["Marks"] >= 80) & (df["Age"] <= 22)
]
print(top_students)
```

```python
islamabad_students = df[df["City"] == "Islamabad"]
print(islamabad_students)
```

```python
result = df[df["Name"].str.contains("a", case=False)]
print(result)
```

```python
# sorting 
print(df.sort_values(by="Marks"))
```

#### Handling missing values

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Fatima"],
    "Age": [21, None, 20, 23],
    "Marks": [85, 92, None, 95]
}

df = pd.DataFrame(data)

print(df)
print(df.isnull())
print(df.isnull().sum())
```

##### Remove rows with missing values

```python
clean_df = df.dropna()
print(clean_df)
```

##### Fill missing values

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print(df)
```

#### Saving DataFrame to csv file

```python
df.to_csv("cleaned_students.csv", index=False)
```
