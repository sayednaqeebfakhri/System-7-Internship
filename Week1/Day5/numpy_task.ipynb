```python
import numpy as np

matrix = np.array([
    [10, 20, 30, 40],
    [ 5, 15, 25, 35],
    [ 2,  4,  6,  8],
    [ 1,  3,  5,  7],
])
print("Original matrix:\n", matrix)
```

```python
print("\nShape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size (total elements):", matrix.size)
print("Data type:", matrix.dtype)
```

```python
# 3. Indexing and slicing
print("\nElement at row 1, col 2:", matrix[1, 2])
```

```python
print("First row:", matrix[0])
```

```python
print("Last column:", matrix[:, -1])
```

```python
print("Top-left 2x2 block:\n", matrix[:2, :2])
```

```python
print("\nElements greater than 20:", matrix[matrix > 20])
print("Even elements:", matrix[matrix % 2 == 0])
```

```python
#  Element-wise math
print("\nMatrix + 5:\n", matrix + 5)
print("Matrix * 2:\n", matrix * 2)
print("Square root:\n", np.sqrt(matrix).round(2))
print("Matrix squared (element-wise):\n", matrix ** 2)
```

```python
#  Aggregate / statistical operations
print("\nSum of all elements:", matrix.sum())
print("Sum of each column:", matrix.sum(axis=0))
print("Sum of each row:", matrix.sum(axis=1))
print("Mean:", matrix.mean())
print("Max:", matrix.max(), "| Min:", matrix.min())
print("Standard deviation:", round(matrix.std(), 2))
print("Index of max value (flattened):", matrix.argmax())
```

```python
# 7. Matrix operations
print("\nTranspose:\n", matrix.T)
print("Matrix multiplication (matrix @ matrix.T):\n", matrix @ matrix.T)
print("Determinant:", round(np.linalg.det(matrix), 2))
```

```python
# 8. Reshaping
print("\nFlattened:", matrix.flatten())
print("Reshaped to 2x8:\n", matrix.reshape(2, 8))
```

```python
# 9. Modifying values
modified = matrix.copy()
modified[0, 0] = 100
modified[modified < 5] = 0
print("\nModified copy:\n", modified)
```
