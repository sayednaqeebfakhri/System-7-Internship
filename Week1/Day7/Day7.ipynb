```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Hours_Studied": [1, 2, 3, 4, 5, 6],
    "Marks": [40, 50, 55, 65, 75, 85],
    "Attendance": [60, 65, 70, 80, 85, 95]
})
```

```python
sns.regplot(data=df, x="Hours_Studied", y="Marks")
plt.title("Study Hours and Marks")
plt.show()
```

```python
sns.boxplot(data=df, y="Marks")
plt.title("Distribution of Marks")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Variables")
plt.show()
```

```python
import numpy as np
import tensorflow as tf

# Training examples
x = np.array([-1, 0, 1, 2, 3, 4], dtype=float)
y = np.array([-3, -1, 1, 3, 5, 7], dtype=float)

# Create a simple neural network
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

# Configure training
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss="mean_squared_error"
)

# Train the model
model.fit(x, y, epochs=200, verbose=0)

# Make a prediction
prediction = model.predict(np.array([10.0]), verbose=0)

print("Prediction for x=10:", prediction[0][0])
```
