#### Conditions (if, elif, else)

```python
# if statement
age = 20

if age >= 18:
    print("You are eligible to vote.")
```

```python
# if , else statement
age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
# if-elif-else statement
marks = 82

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Needs improvement")
```

```python
# Nested conditions
age = 22
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Entry denied")
```

```python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
```

#### Loops in Python

```python
# for loop
for i in range(5):
    print(i)
```

```python
names = ["Ali", "Sara", "Ahmed"]

for name in names:
    print(name)
```

```python
# while loop
count = 0
while count < 5:
    print(count)
    count += 1
```

##### break and continue

```python
# break statement
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
# continue statement

for i in range(1, 10):
    if i == 5:
        continue
    print(i)
```

```python
# range() function
for i in range(5):
    print(i)
```

```python
# range(start, stop)
for i in range(2, 10):
    print(i)
```

```python
# range(start, stop, step)

for i in range(2, 11, 2):
    print(i)
```

#### Functions in Python

```python
def greet():
    print("Welcome to Python!")

greet()
greet()
```

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Naqeeb")
```

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

```python
# Functions with default parameters
def greet(name="Student"):
    print(f"Hello, {name}!")

greet()
greet("Naqeeb")
```

```python
# function with conditional logic

def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(10))
print(check_even_odd(7))
```

```python
def calculate_average(marks):
    return sum(marks) / len(marks)


def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    else:
        return "Needs improvement"


name = input("Enter student name: ")

marks = []

for i in range(1, 4):
    mark = float(input(f"Enter marks for subject {i}: "))
    marks.append(mark)

total = sum(marks)
average = calculate_average(marks)
grade = calculate_grade(average)

print("\n--- Student Result ---")
print(f"Name: {name}")
print(f"Total: {total}")
print(f"Average: {average:.2f}")
print(f"Grade: {grade}")
```
