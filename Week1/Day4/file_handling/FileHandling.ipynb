#### Opening a file

```python
file = open("notes.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
with open("notes.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
with open("notes.txt", "r") as file:
    content = file.read(4)
    print(content)
```

```python
with open("notes.txt", "r") as file1:
    print(file1.read(4))
    print(file1.read(4))
    print(file1.read(7))
    print(file1.read(15))
```

```python

with open("notes.txt", "r") as file1:
    print("first line: " + file1.readline())
    print("second line: " + file1.readline())
```

```python
with open("notes.txt", "r") as file1:
    for line in file1:
        print(line, end="")
```

```python

with open("notes.txt", "r") as file1:
    FileasList = file1.readlines()
print(FileasList[0])
```

#### Appending 
The write method overwrites the content in a file.
the append method adds new lines  and doesn't remove the previous lines


```python
with open("notes.txt", "a") as file:
    file.write("\nAdding a new line to the file.")
    file.write("\nAdding another line to the file.")
```

```python
with open("notes.txt", "r") as file:
    for line in file:
        print(line, end="")
```

```python
with open("notes2.txt", "a+") as file:
    file.write("\nAdding a new line to notes2.txt.")
    file.write("\nAdding another line to notes2.txt.")
```

```python
with open("notes.txt", "r") as file:
    with open("notes2.txt", "a") as file2:
        for line in file:
            file2.write(line)
```

```python
with open("notes2.txt", "r") as file:
    content = file.read()
    print(content)
```
