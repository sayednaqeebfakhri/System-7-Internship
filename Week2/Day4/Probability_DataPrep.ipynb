#### Probability and Data Preparation

Probability
Probability measures how likely an event is to happen. Its value ranges from 0 to 1, where 0 means impossible and 1 means certain.

P(A) =  Favorable outcomes / Total outcomes

For example, when rolling a fair six-sided die, the probability of getting a 3 is \(1/6\), because there is one favorable outcome out of six possible outcomes.

In Machine Learning, probability is useful for estimating the likelihood of outcomes, such as whether a student might pass a course.

```python
# Probability of passing if 75 out of 100 students pass
passed = 75
total_students = 100

probability = passed / total_students

print("Probability of passing:", probability)
print("Percentage:", probability * 100, "%")
```

**Correlation**

Correlation measures the strength and direction of a relationship between two numerical variables. Its correlation coefficient usually ranges from -1 to +1.


**Positive correlation**

As one variable increases, the other tends to increase. 
Example: study hours and marks may have a positive correlation.


**Negative correlation**

As one variable increases, the other tends to decrease. Example: travel speed and travel time for a fixed distance.


**Zero or near-zero linear correlation**

There is little or no linear relationship between the variables. A nonlinear relationship may still exist.



The Pearson correlation coefficient is commonly represented by r. Values close to +1 indicate a strong positive linear relationship, values close to -1 indicate a strong negative linear relationship, and values near 0 indicate a weak linear relationship.

```python
import pandas as pd

data = {
    "StudyHours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

print(df["StudyHours"].corr(df["Marks"]))
```

**Train, Validation, and Test Split**

In Machine Learning, we divide a dataset into separate parts so we can train a model and evaluate how well it works on unseen data

```python
from sklearn.model_selection import train_test_split

X = list(range(100))  # Example feature values
y = [value * 2 for value in X]  # Example targets

# First: reserve 15% for testing
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42
)

# Second: reserve 15% of the original dataset for validation
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp,
    test_size=0.15 / 0.85,
    random_state=42
)

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Testing samples:", len(X_test))
```
